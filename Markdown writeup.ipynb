## The expression below gives the cross section of the processes $e^{+}e^{-}\rightarrow  \mu^+\mu^-$, as a function of the centre of mass energy $\sqrt{s}$. The parameters $m_{Z}$, $\Gamma_{Z}$ denote the mass and width of the $Z^0$ boson respectively ##

$$\sigma(e^{+}e^{-}\rightarrow  \mu^+\mu^-) = \frac{12\pi s}{m_{Z}^{2}}\frac{\Gamma_{e^+e^-}\Gamma_{\mu^+\mu^-}}{(s-m_{Z}^{2})^{2}+m_{Z}^{2}\Gamma_{Z}^{2}}$$

# Question 1 [8 marks] #

- What is the key assumption regarding the range of $\sqrt{s}$ values for which the above expression holds?
- What is the physical meaning of $\Gamma_Z$?
- What is the physical meaning of $\Gamma_{e^+e^-}$ and $\Gamma_{\mu^+\mu^-}$?

1. This equation represents the cross section for the proceses of an electron-positron collision producing a muon-antimuon pair, via a $Z^0$ boson: $e^{+}e^{-}\rightarrow Z^0 \rightarrow \mu^+\mu^-$. The main assumption here is that the centre of mass energy is close to the $Z^0$ resonance ($\sqrt{s} \approx m_Z$), such that this process dominates over other processes. $e^{+}e^{-}\rightarrow \gamma \rightarrow \mu^+\mu^-$, for example, would dominates at lower energies, and more complex processes which dominate at higher energies.

2. $\Gamma_Z$ represents the decay width of the $Z^0$ boson. It is realted to the mean lifetime of the $Z^0$, $\tau_{Z}$ by 

$$\tau_Z = \hbar / \Gamma_Z$$

3. $\Gamma_{e^+e^-}$ and $\Gamma_{\mu^+\mu^-}$ represent the partial widths of the processes by which the $Z_0$ boson decays either to a $e^+$, $e^-$ pair or a $\mu^+$, $\mu^-$ pair respectively. The total width is the sum of all of the partial widths ($Z_0$ boson decay channels) by $\Gamma_{TOTAL} = \sum_{i} \Gamma_i = \Gamma_{e^+e^-} + \Gamma_{\mu^+\mu^-} + ...$ 



# Question 2 [4 marks] # 
- Make a plot of the cross-section expression given above as a function of the centre of mass energy $\sqrt{s}$ in the range between 0.210GeV and 120GeV, using the following values of the parameters $m_Z$, $\Gamma_Z$, $\Gamma_{e^+e^-}$, $\Gamma_{\mu^+\mu^-}$:\
$m_Z=91.18$GeV\
$\Gamma_Z=2.496$GeV\
$\Gamma_{e^+e^-}=\Gamma_{\mu^+\mu^-}=0.084$GeV

```python
import numpy as np
import matplotlib.pyplot as plt

def sigma(sqrt_s, mass_Z, gamma_Z, gamma_ee, gamma_mumu):
    sigma = (12*np.pi* sqrt_s**2) / (mass_Z**2) * (gamma_ee * gamma_mumu) / ( (sqrt_s**2 - mass_Z**2)**2 + mass_Z**2 * gamma_Z**2)
    return sigma

m_Z, g_Z = 91.18, 2.496 # in GeV
g_ee, g_mumu = 0.084, 0.084 # in GeV
sqrt_s= np.linspace(0.210, 120, 1000) # in GeV

plt.figure(figsize=(9, 6))
plt.plot(sqrt_s,sigma(sqrt_s, m_Z, g_Z, g_ee, g_mumu))
plt.xlabel(r"$\sqrt{s}$ [GeV]")
plt.ylabel(r"$\sigma(e^+e^-\rightarrow\mu^+\mu^-)$ [GeV$^{-2}$]")
plt.grid()
plt.show()
```

- Devise an approach to determine the number of generations of neutrinos by performing a fit to the histogram you created in the previous cell. 
    - You are given that in the SM: $\Gamma_{\nu_{l}\bar{\nu}_{l}}=2\Gamma_{l^+l^-}$, where $l$ denotes a specific lepton generation
    - You can also assume the values: $\Gamma_{q\bar{q}}=1.744$GeV, $\Gamma_{e^+e^-}=\Gamma_{\mu^+\mu^-}=\Gamma_{\tau^+\tau^-}=0.083984$GeV

**Hint: Take a look at particle_project_example.ipynb for an example of how to fit a function to a histogram. Make sure you also look at how to fix a parameter so that it is not varying in the fit**

We know that $\Gamma_Z = \Sigma_F\Gamma(Z^0\rightarrow F)$ or $$\Gamma_Z = \Gamma_{q\bar{q}} + 3\Gamma_{l^+l^-} + N_\nu\Gamma_{\nu\bar{\nu}}$$ Where N_\nu is the number of neutrino generations. Therefore: $$\Gamma_Z = 1.744 + (0.083984)(3+2N_\nu)$$ $$\Gamma_Z = 1.995952 +0.167968N_\nu$$ we can substitute this into the equation for the cross section: $$\sigma(e^{+}e^{-}\rightarrow  \mu^+\mu^-) = \frac{12\pi s}{m_{Z}^{2}}\frac{\Gamma_{e^+e^-}\Gamma_{\mu^+\mu^-}}{(s-m_{Z}^{2})^{2}+m_{Z}^{2}(1.995952 +0.167968N_\nu)^{2}}$$ and fit to find $N_\nu$. Given our earlier value of $\Gamma_Z = 2.496GeV$ we should expect $N_\nu \approx 2.977$

```python
Gamma_Z = 2.496

Gamma_ee = 0.083984
Gamma_mumu = 0.083984
Gamma_tautau = 0.083984 

Gamma_had = 1.744
Gamma_lepton = Gamma_mumu + Gamma_ee + Gamma_tautau

#------------------------------------------------------
# Step 1: create equation 1 and equation 2: total width and cross section
#------------------------------------------------------
def Gamma_Z_nu(N_nu):
    return Gamma_had + Gamma_lepton + 2 * N_nu * Gamma_mumu

def sigma(sqrt_s, N_nu, m_Z = 91.18):
    s = sqrt_s**2
    Gamma_tot = Gamma_Z_nu(N_nu)

    sigma = ( 12 * np.pi * s / m_Z**2) * (Gamma_lepton**2) / ((s - m_Z**2)**2 + m_Z**2 * Gamma_tot**2) 

    return sigma

print(Gamma_Z_nu(3)) # This print 2.499856, different to the previos gammas used 2.496



#------------------------------------------------------
# Step 2: Calculate the CDF using func_cdf from utils.py
#------------------------------------------------------

def cdf_nu(bin_edges, N_nu, m_Z):
    return func_cdf(sigma, bin_edges, N_nu, m_Z)



#------------------------------------------------------
# Step 3: Fit the histogram using Minuit
#------------------------------------------------------

# Binned likelihood
cost = BinnedNLL(counts, bin_edges, cdf_nu)

# Start the fit with 3 neutrino generations
fit = Minuit(cost, N_nu=3.0, m_Z=91.18)

# Fix the Z mass to the value used when generating the events
fit.fixed["m_Z"] = True

# N_nu cannot be negative
fit.limits["N_nu"] = (0, None)


# Run the fit and uncertainties
fit.migrad()
fit.hesse()



# Show the results
display(fit)
```

```python
# Extract the number of neutrino generations and uncertainty
N_nu_fit = fit.values["N_nu"]
N_nu_err = fit.errors["N_nu"]

Gamma_Z_fit = Gamma_Z_nu(N_nu_fit)
Gamma_Z_err = 2 * Gamma_lepton * N_nu_err

print(f"N_nu = {N_nu_fit:.5f} +/- {N_nu_err:.5f}")
print(f"Gamma_Z = {Gamma_Z_fit:.6f} +/- {Gamma_Z_err:.6f} GeV")
```

```python
#------------------------------------------------------
# Step 4: Compare the fit with the generated data
#------------------------------------------------------

# Get the centre of each bin
bin_centres = (bin_edges[:-1] + bin_edges[1:]) / 2

# Calculate the expected events using CDF
cdf_values = cdf_nu(bin_edges, N_nu_fit, 91.18)
expected_counts = np.sum(counts) * np.diff(cdf_values)

# Plot the generated data and the fit
plt.figure(figsize=(9, 6))
plt.errorbar(bin_centres, counts, yerr=np.sqrt(counts), fmt=".", label="Generated data", markersize=8, elinewidth=1)
plt.plot(bin_centres, expected_counts, label=rf"Fit: $N_\nu={N_nu_fit:.4f}\pm{N_nu_err:.4f}$")
plt.xlabel(r"$\sqrt{s}$ [GeV]")
plt.ylabel("Number of events")
plt.title(r"$Z^0$ resonance fit")
plt.legend(loc="upper left", fontsize=16)
plt.show()
```

```python
from scipy.stats import chi2
from IPython.display import display, Math

chi2_value = np.sum((counts - expected_counts)**2 / expected_counts)

ndof = len(counts) - 1 - 1  # 1 normalisation constraint, 1 fitted parameter
p_value = chi2.sf(chi2_value, ndof)

display(Math(rf"\text{{Reduced }} \chi^2 = {chi2_value / ndof:.3f}"))

print(f"p-value = {p_value:.4f}")
```

2. The systematic uncertaintaies associated with the inputs of question 7 are the uncertainties on: $\Gamma_Z$, $\Gamma_{q\bar{q}}$, $\Gamma_{l^+l^-}$, and $m_Z$ Their uncertainties according to the Particle Data Group are: $$\Delta\Gamma_Z = 0.0023GeV$$ $$\Delta\Gamma_{q\bar{q}} = 0.0020GeV$$ $$\Delta\Gamma_{l^+l^-} = 0.000086GeV$$ $$\Delta m_Z = 0.0020 GeV$$ The partial widths for the electron and muon pairs do not effect the uncertainty in the crossection fit, because their contribution disappears when the fit is normalised, however they do contribute uncertainty through their relation to $\Gamma_Z$ and the assumed $\Gamma_{\nu\bar{\nu}}$. However, this contribution cannot be directly added as we do not use experimental luminoscity measurements. 

3. to estimate the effect of the systematic uncertainties on our value of $N_\nu$, we propogate the errors through the equation from question 7: $$N_\nu = \frac{\Gamma_Z - \Gamma_{q\bar{q}} - 3\Gamma_{l^+l^-}}{2\Gamma_{l^+l^-}}$$

$$\Delta N_\nu = \sqrt{\left(\frac{\partial N_\nu}{\partial\Gamma_{q\bar{q}}}\Delta \Gamma_{q\bar{q}}\right)^2 + \left(\frac{\partial N_\nu}{\partial\Gamma_{l^+l^-}}\Delta\Gamma_{l^+l^-}\right)^2 +\left(\frac{\partial N_\nu}{\partial\Gamma_{Z}}\Delta\Gamma_{Z}\right)^2}$$

where $$\frac{\partial N_\nu}{\partial \Gamma_{q\bar{q}}}=-\frac{1}{2\Gamma_{l^+l^-}}$$ $$\frac{\partial N_\nu}{\partial \Gamma_{l^+l^-}}=\frac{-\Gamma_Z - \Gamma_{q\bar{q}}}{2\Gamma_{l^+l^-}^2}$$ $$\frac{\partial N_\nu}{\partial \Gamma_{Z}}=\frac{1}{2\Gamma_{l^+l^-}}$$ gives $$\Delta N_\nu = 0.01872$$

meaning our result now only deviates from $3$ by $1.18\sigma$ - a much better statistical compatibility.

However, during the fitting process, both the values of $\Gamma_{q\bar{q}}$ and $\Gamma_{l^+l^-}$ are held constant, and the final value of error given for the fit $(\Delta N_\nu)$ is therefore directly dependent on $\Gamma_Z$. This error may therefore not be treated as an indepentent systematic uncertainty, and instead my be replaced by the statistical uncertainty from the fit. The total sytematic uncertainty then is $$\Delta N_\nu^{SYST} = \sqrt{\left(\frac{\partial N_\nu}{\partial\Gamma_{q\bar{q}}}\Delta \Gamma_{q\bar{q}}\right)^2 + \left(\frac{\partial N_\nu}{\partial\Gamma_{l^+l^-}}\Delta\Gamma_{l^+l^-}\right)^2}$$ Recalling that the error from the fit $(\Delta N_\nu^{FIT})$ is $0.00556$ and summing the errors: $$N_\nu = 2.97826\pm 0.01391$$ This yeilds a statistical variation from the expected value of $3$ of $1.56\sigma$ - slightly less statistically compatible then the previous calculation, but still significantly better than that yeilded by the fit error only.
