## The expression below gives the cross section of the processes $e^{+}e^{-}\rightarrow  \mu^+\mu^-$, as a function of the centre of mass energy $\sqrt{s}$. The parameters $m_{Z}$, $\Gamma_{Z}$ denote the mass and width of the $Z^0$ boson respectively ##

$$\sigma(e^{+}e^{-}\rightarrow  \mu^+\mu^-) = \frac{12\pi s}{m_{Z}^{2}}\frac{\Gamma_{e^+e^-}\Gamma_{\mu^+\mu^-}}{(s-m_{Z}^{2})^{2}+m_{Z}^{2}\Gamma_{Z}^{2}}$$

# Question 1 [8 marks] #

- What is the key assumption regarding the range of $\sqrt{s}$ values for which the above expression holds?
- What is the physical meaning of $\Gamma_Z$?
- What is the physical meaning of $\Gamma_{e^+e^-}$ and $\Gamma_{\mu^+\mu^-}$?

1. This equation represents the cross section for the process in which an electron and a positron collide producing a muon-antimuon pair, via a $Z^0$ boson: $e^{+}e^{-}\rightarrow Z^0 \rightarrow \mu^+\mu^-$. The main assumption is that the centre-of-mass energy is close to the $Z^0$ resonance ($\sqrt{s}\approx m_Z$), where the $Z^0$ contribution dominates. The equation does not account for virtual-photon exchange ($e^+e^-\rightarrow\gamma^*\rightarrow\mu^+\mu^-$) and its interference with the $Z^0$ contribution. These effects would become more significant at energies away from the resonance.


2. $\Gamma_Z$ represents the decay width of the $Z^0$ boson. It is related to the mean lifetime of the $Z^0$, $\tau_{Z}$ by 

$$\tau_Z = \hbar / \Gamma_Z$$

3. $\Gamma_{e^+e^-}$ and $\Gamma_{\mu^+\mu^-}$ represent the partial widths of the processes by which the $Z^0$ boson decays either into an $e^+$ $e^-$ pair or a $\mu^+$, $\mu^-$ pair respectively. The total width is the sum of all of the partial widths ($Z^0$ boson decay channels) by $\Gamma_{TOTAL} = \sum_{i} \Gamma_i = \Gamma_{e^+e^-} + \Gamma_{\mu^+\mu^-} + ...$ 



# Question 2 [4 marks] # 
- Make a plot of the cross-section expression given above as a function of the centre of mass energy $\sqrt{s}$ in the range between 0.210GeV and 120GeV, using the following values of the parameters $m_Z$, $\Gamma_Z$, $\Gamma_{e^+e^-}$, $\Gamma_{\mu^+\mu^-}$:\
$m_Z=91.18$GeV\
$\Gamma_Z=2.496$GeV\
$\Gamma_{e^+e^-}=\Gamma_{\mu^+\mu^-}=0.084$GeV

```python
import numpy as np
import matplotlib.pyplot as plt


def sigma(sqrt_s, mass_Z, gamma_Z, gamma_ee, gamma_mumu):
    sigma = (12*np.pi* sqrt_s**2) / (mass_Z**2) * (gamma_ee * gamma_mumu) / ( (sqrt_s**2 - mass_Z**2)**2 + mass_Z**2 * gamma_Z**2)
    return sigma

m_Z, g_Z = 91.18, 2.496 # in GeV
g_ee, g_mumu = 0.084, 0.084 # in GeV
sqrt_s= np.linspace(0.210, 120, 1000) # in GeV

plt.figure(figsize=(9, 6))
plt.plot(sqrt_s,sigma(sqrt_s, m_Z, g_Z, g_ee, g_mumu))
plt.xlabel(r"$\sqrt{s}$ [GeV]")
plt.ylabel(r"$\sigma(e^+e^- \rightarrow \mu^+ \mu^-)$ [GeV$^{-2}$]")
plt.grid()
plt.show()
```

From the previous question we know that $\Gamma_Z=2.496GeV$. $$BF(I\rightarrow F) = \frac{\Gamma(I\rightarrow F)}{\Gamma_{TOTAL}}$$ Given that $BF(Z^0 \rightarrow e^+e^-) = 3.366%$ and $BF(Z^0 \rightarrow \mu^+\mu^-) = 3.366%$ we can calculate: $$\Gamma_{e^+e^-} = BF(Z^0\rightarrow e^+e^-)\Gamma_Z$$ $$ = 0.03366 \times 2.496 GeV$$ $$ = 0.084GeV$$ Similarly for the muon pair: $$\Gamma_{\mu^+\mu^-} = BF(Z^0\rightarrow \mu^+\mu^-)\Gamma_Z$$ $$ = 0.03366 \times 2.496 GeV$$ $$ = 0.084GeV$$


2. The cross section function should not simply decrease as the centre-of-mass energy decreases from the $Z^0$ resonance. At lower energies, the process can also occur through a virtual photon as a mediator, $e^+e^-\rightarrow\gamma^*\rightarrow\mu^+\mu^-$. This contribution becomes important away from the $Z^0$ resonance. However, the physical cross section must approach zero at the two muons production threshold, $\sqrt{s}=2m_\mu\approx0.2113$ GeV (the minimum centre-of-mass energy needed to produce a muon pair), because the available phase space vanishes. The plot made in question 2 shows only the $Z^0$ contribution and therefore does not describe other types of contributions for lower energies.

# Question 7 [28 marks] #
- Devise an approach to determine the number of generations of neutrinos by performing a fit to the histogram you created in the previous cell. 
    - You are given that in the SM: $\Gamma_{\nu_{l}\bar{\nu}_{l}}=2\Gamma_{l^+l^-}$, where $l$ denotes a specific lepton generation
    - You can also assume the values: $\Gamma_{q\bar{q}}=1.744$GeV, $\Gamma_{e^+e^-}=\Gamma_{\mu^+\mu^-}=\Gamma_{\tau^+\tau^-}=0.083984$GeV

**Hint: Take a look at particle_project_example.ipynb for an example of how to fit a function to a histogram. Make sure you also look at how to fix a parameter so that it is not varying in the fit**

We know that $\Gamma_Z = \Sigma_F\Gamma(Z^0\rightarrow F)$ or $$\Gamma_Z = \Gamma_{q\bar{q}} + 3\Gamma_{l^+l^-} + N_\nu\Gamma_{\nu\bar{\nu}}$$ Where N_\nu is the number of neutrino generations. Therefore: $$\Gamma_Z = 1.744 + (0.083984)(3+2N_\nu)$$ $$\Gamma_Z = 1.995952 +0.167968N_\nu$$ we can substitute this into the equation for the cross section: $$\sigma(e^{+}e^{-}\rightarrow  \mu^+\mu^-) = \frac{12\pi s}{m_{Z}^{2}}\frac{\Gamma_{e^+e^-}\Gamma_{\mu^+\mu^-}}{(s-m_{Z}^{2})^{2}+m_{Z}^{2}(1.995952 +0.167968N_\nu)^{2}}$$ and fit to find $N_\nu$. Given our earlier value of $\Gamma_Z = 2.496GeV$ we should expect $N_\nu \approx 2.977$

```python
Gamma_Z = 2.496

Gamma_ee = 0.083984
Gamma_mumu = 0.083984
Gamma_tautau = 0.083984 

Gamma_had = 1.744
Gamma_lepton = Gamma_mumu + Gamma_ee + Gamma_tautau
Gamma_l = Gamma_lepton/3 # Width for one charged lepton
#------------------------------------------------------
# Step 1: create equation 1 and equation 2: total width and cross section
#------------------------------------------------------
def Gamma_Z_nu(N_nu):
    return Gamma_had + Gamma_lepton + 2 * N_nu * Gamma_l

def sigma_nu(sqrt_s, N_nu, m_Z = 91.18):
    s = sqrt_s**2
    Gamma_tot = Gamma_Z_nu(N_nu)

    sigma = (12 * np.pi * s / m_Z**2) * (Gamma_ee * Gamma_mumu) / ((s - m_Z**2)**2 + m_Z**2 * Gamma_tot**2)

    return sigma

print(Gamma_Z_nu(3)) # This prints 2.499856, different to the previous total width



#------------------------------------------------------
# Step 2: Calculate the CDF using func_cdf from utils.py
#------------------------------------------------------

def cdf_nu(bin_edges, N_nu, m_Z):
    return func_cdf(sigma_nu, bin_edges, N_nu, m_Z)



#------------------------------------------------------
# Step 3: Fit the histogram using Minuit
#------------------------------------------------------

# Binned likelihood
cost = BinnedNLL(counts, bin_edges, cdf_nu)

# Start the fit with 3 neutrino generations
fit = Minuit(cost, N_nu=3.0, m_Z=91.18)

# Fix the Z mass to the value used when generating the events
fit.fixed["m_Z"] = True

# N_nu cannot be negative
fit.limits["N_nu"] = (0, None)


# Run the fit 
fit.migrad()
fit.hesse()

display(fit)
```

```python
# Extract the number of neutrino generations and uncertainty
N_nu_fit = fit.values["N_nu"]
N_nu_err = fit.errors["N_nu"]

Gamma_Z_fit = Gamma_Z_nu(N_nu_fit)
Gamma_Z_err = 2 * Gamma_l * N_nu_err

print(f"N_nu = {N_nu_fit:.5f} +/- {N_nu_err:.5f}")
print(f"Gamma_Z = {Gamma_Z_fit:.6f} +/- {Gamma_Z_err:.6f} GeV")
```

```python
#------------------------------------------------------
# Step 4: Compare the fit with the generated data
#------------------------------------------------------

# Get the centre of each bin
bin_centres = (bin_edges[:-1] + bin_edges[1:]) / 2

# Calculate the expected events using CDF
cdf_values = cdf_nu(bin_edges, N_nu_fit, 91.18)
expected_counts = np.sum(counts) * np.diff(cdf_values)

# Plot the generated data and the fit
plt.figure(figsize=(9, 6))
plt.errorbar(bin_centres, counts, yerr=np.sqrt(counts), fmt=".", label="Generated data", markersize=8, elinewidth=1)
plt.plot(bin_centres, expected_counts, label=rf"Fit: $N_\nu={N_nu_fit:.4f}\pm{N_nu_err:.4f}$")
plt.xlabel(r"$\sqrt{s}$ [GeV]")
plt.ylabel("Number of events")
plt.title(r"$Z^0$ resonance fit")
plt.legend(loc="upper left", fontsize=16)
plt.show()
```

```python
from scipy.stats import chi2
from IPython.display import display, Math

chi2_value = np.sum((counts - expected_counts)**2 / expected_counts)

ndof = len(counts) - 1 - 1  # 1 normalisation constraint, 1 fitted parameter
p_value = chi2.sf(chi2_value, ndof)

display(Math(rf"\text{{Reduced }} \chi^2 = {chi2_value / ndof:.3f}"))

print(f"p-value = {p_value:.4f}")
```

1. Our fit gives $N_\nu=2.97826\pm0.00556$. This differs from three generations of neutrinos by $3.91\sigma$.

    $$z=\frac{|2.97826-3|}{0.00556}\approx3.91. $$

    This is not statistically compatible with three generations at the 95% confidence level. One reason may be the difference between the value of $\Gamma_Z = 2.496 GeV$ used to generate the random data and the original PDG value plus random errors. It should be notice that we got a reduced chi-squared of $0.996$ and p-value of approximately $0.4998$, which indicate that the fitted model describes the simulated data well; however, a good fit does not necessarily imply compatibility with $N_\nu=3$.




2. The main systematic uncertainties are associated with the hadronic partial width, the charged lepton partial widths, and the fixed $Z^0$ mass that we used in Question 7.

    The Particle Data Group (PDG) [2] reports the following experimental uncertainties:

    $\Gamma_Z$, $\Gamma_{q\bar{q}}$, $\Gamma_{l^+l^-}$, and $m_Z$ Their uncertainties according to the Particle Data Group are: $$\Delta\Gamma_Z = 0.0023GeV$$ $$\Delta\Gamma_{q\bar{q}} = 0.0020GeV$$ $$\Delta\Gamma_{l^+l^-} = 0.000086GeV$$ $$\Delta m_Z = 0.0020 GeV$$

    The partial widths for the electron and muon pairs do not affect the uncertainty in the crossection fit, because their product cancels when the distribution is normalised for the likelihood fit; however, they do contribute uncertainty through their relation to $\Gamma_Z$ and the assumed $\Gamma_{\nu\bar{\nu}}$.

    In our model, we assume lepton universality: $ \Gamma_{ee}=\Gamma_{\mu\mu}=\Gamma_{\tau\tau} =\Gamma_{\ell^+\ell^-}. $ Mele (2015) [1] states that the ratio between $\Gamma_{\nu\bar{\nu}}$ and $\Gamma_{l^+l^-}$ is $1.99125\pm 0.00083$ which would produce a slight error in our result. That's why another source of uncertainty is the approximation used in Question 7: $ \frac{\Gamma_{\nu\bar\nu}}{\Gamma_{\ell^+\ell^-}}=2.$ Other systematic effects include the uncertainty in the fixed $Z^0$ mass and the neglected virtual-photon exchange and $\gamma-Z$ interference contributions. Mele [1] also identifies luminosity determination through low angle scattering as a major source of uncertainty. However, this contribution cannot be directly added as we do not use experimental luminosity measurements. 








3. To estimate the effect of the systematic uncertainties on our value of $N_\nu$, we propagate the errors through the equation from Question 7:

    $$
    N_\nu = \frac{\Gamma_Z - \Gamma_{q\bar{q}} - 3\Gamma_{l^+l^-}}{2\Gamma_{l^+l^-}}
    $$

    Assuming the input uncertainties are independent,

    $$
    \Delta N_\nu = \sqrt{\left(\frac{\partial N_\nu}{\partial\Gamma_{q\bar{q}}}\Delta \Gamma_{q\bar{q}}\right)^2 + \left(\frac{\partial N_\nu}{\partial\Gamma_{l^+l^-}}\Delta\Gamma_{l^+l^-}\right)^2 + \left(\frac{\partial N_\nu}{\partial\Gamma_{Z}}\Delta\Gamma_{Z}\right)^2}
    $$

    where

    $$
    \frac{\partial N_\nu}{\partial \Gamma_{q\bar{q}}}=-\frac{1}{2\Gamma_{l^+l^-}}
    $$

    $$
    \frac{\partial N_\nu}{\partial \Gamma_{l^+l^-}}=-\frac{\Gamma_Z - \Gamma_{q\bar{q}}}{2\Gamma_{l^+l^-}^2}
    $$

    $$
    \frac{\partial N_\nu}{\partial \Gamma_{Z}}=\frac{1}{2\Gamma_{l^+l^-}}
    $$

    Using the uncertainties from the PDG gives

    $$
    \Delta N_\nu^{PDG} = \sqrt{0.01191^2+0.00459^2+0.01369^2}\approx0.01872
    $$

    However, this is an estimate using external PDG uncertainties, rather than the uncertainty of our simulated fit. The PDG measurements may also be correlated, which is not considered here.

    During the fitting process, both $\Gamma_{q\bar{q}}$ and $\Gamma_{l^+l^-}$ are held constant, and the statistical uncertainty obtained from the fit is therefore related to the uncertainty in the fitted $\Gamma_Z$. We should not add the external PDG uncertainty in $\Gamma_Z$ as an independent systematic uncertainty, since the uncertainty in our fitted $\Gamma_Z$ is already represented by the Minuit statistical uncertainty in $N_\nu$.

    The systematic uncertainty from the two input partial widths is then

    $$
    \Delta N_\nu^{SYST} = \sqrt{\left(\frac{\partial N_\nu}{\partial\Gamma_{q\bar{q}}}\Delta \Gamma_{q\bar{q}}\right)^2 + \left(\frac{\partial N_\nu}{\partial\Gamma_{l^+l^-}}\Delta\Gamma_{l^+l^-}\right)^2}
    $$

    $$
    \Delta N_\nu^{SYST}=\sqrt{0.01191^2+0.00459^2}\approx0.01276
    $$

    Recalling that the statistical uncertainty from the fit is $\Delta N_\nu^{FIT}=0.00556$, we combine the uncertainties:

    $$
    \Delta N_\nu^{TOTAL}=\sqrt{0.00556^2+0.01276^2}\approx0.01392
    $$

    This gives

    $$
    N_\nu = 2.97826\pm0.00556\;(\mathrm{stat})\pm0.01276\;(\mathrm{syst})
    $$

    The discrepancy from the expected value of three generations is then

    $$
    z=\frac{|2.97826-3|}{0.01392}\approx1.56
    $$

    This reduces the estimated discrepancy to $1.56\sigma$, making the result compatible with three generations at the 95% confidence level under these assumptions.

    Another way of estimating the systematic effects would be to vary each input parameter by its PDG uncertainty and re-fit. For instance we could vary the fixed $Z^0$ mass or use the more precise neutrino width ratio of $1.99125\pm0.00083$ from Mele (2015). This would tell us how sensitive our result is to the assumptions of the model.
